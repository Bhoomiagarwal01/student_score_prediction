```python
import pandas as pd
from sklearn.linear_model import LinearRegression

data = pd.DataFrame({
    "Hours":[2,4,6,8,10],
    "Score": [15,30,45,60,75]
})

X = data[["Hours"]]
y = data["Score"]

model = LinearRegression()
model.fit(X, y)

# ✅ Fix: use DataFrame with same column name
new_data = pd.DataFrame({"Hours": [6]})
print(model.predict(new_data))
import matplotlib.pyplot as plt
plt.scatter(data["Hours"],data["Score"])
plt.show
```
